```python
# ============================================
# FEEDFORWARD NEURAL NETWORK ON CIFAR-10
# KERAS + TENSORFLOW
# ============================================

# Step 1: Import required libraries

import numpy as np
import matplotlib.pyplot as plt

from tensorflow.keras.datasets import cifar10
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import SGD
from tensorflow.keras.utils import to_categorical
from sklearn.metrics import classification_report


# ============================================
# Step 2: Load CIFAR-10 dataset
# ============================================

print("[INFO] Loading CIFAR-10 dataset...")

(trainX, trainY), (testX, testY) = cifar10.load_data()


# ============================================
# Step 3: Preprocess the images
# ============================================

# Flatten 32 x 32 x 3 images into 3072-dimensional vectors

trainX = trainX.reshape(trainX.shape[0], 32 * 32 * 3)
testX = testX.reshape(testX.shape[0], 32 * 32 * 3)

# Convert pixel values to float

trainX = trainX.astype("float32")
testX = testX.astype("float32")

# Normalize pixel values from [0, 255] to [0, 1]

trainX = trainX / 255.0
testX = testX / 255.0

print("Training data shape:", trainX.shape)
print("Testing data shape:", testX.shape)


# ============================================
# Step 4: One-hot encode the labels
# ============================================

trainY = to_categorical(trainY, 10)
testY = to_categorical(testY, 10)

print("Training labels shape:", trainY.shape)
print("Testing labels shape:", testY.shape)


# ============================================
# Step 5: Define the Feedforward Neural Network
# ============================================

model = Sequential()

# Input layer + first hidden layer
model.add(
    Dense(128, input_shape=(3072,), activation="relu")
)

# Second hidden layer
model.add(
    Dense(128, activation="relu")
)

# Output layer - 10 classes
model.add(
    Dense(10, activation="softmax")
)


# ============================================
# Step 6: Compile the model using SGD
# ============================================

sgd = SGD(learning_rate=0.01)

model.compile(
    loss="categorical_crossentropy",
    optimizer=sgd,
    metrics=["accuracy"]
)


# Display model architecture

model.summary()


# ============================================
# Step 7: Train the model
# ============================================

print("[INFO] Training the network...")

H = model.fit(
    trainX,
    trainY,
    validation_data=(testX, testY),
    epochs=20,
    batch_size=128,
    verbose=1
)


# ============================================
# Step 8: Evaluate the model
# ============================================

print("[INFO] Evaluating the network...")

loss, accuracy = model.evaluate(
    testX,
    testY,
    verbose=0
)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)


# ============================================
# Step 9: Generate predictions
# ============================================

predictions = model.predict(
    testX,
    batch_size=128
)

predicted_classes = predictions.argmax(axis=1)
actual_classes = testY.argmax(axis=1)


# ============================================
# Step 10: Classification report
# ============================================

class_names = [
    "airplane",
    "automobile",
    "bird",
    "cat",
    "deer",
    "dog",
    "frog",
    "horse",
    "ship",
    "truck"
]

print("\nClassification Report:")

print(
    classification_report(
        actual_classes,
        predicted_classes,
        target_names=class_names
    )
)


# ============================================
# Step 11: Plot training loss
# ============================================

plt.figure()

plt.plot(
    H.history["loss"],
    label="Training Loss"
)

plt.plot(
    H.history["val_loss"],
    label="Validation Loss"
)

plt.title("Training and Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.legend()

plt.show()


# ============================================
# Step 12: Plot training accuracy
# ============================================

plt.figure()

plt.plot(
    H.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    H.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.title("Training and Validation Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.legend()

plt.show()
```
